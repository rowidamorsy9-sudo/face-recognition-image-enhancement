```python



folder = "/content/dataset/train"


all_files = [f for f in os.listdir(folder) if f.endswith(".jpg")]


def get_base(filename):
    return ''.join([c for c in filename if not c.isdigit()]).replace('_','').replace('.jpg','')


unique_bases = list(set(map(get_base, all_files)))
class_names = {name: idx for idx, name in enumerate(unique_bases)}


labels = list(map(lambda f: (f, class_names[get_base(f)]), all_files))

target_class = 'Fahim...'
class_id = class_names[target_class]


images_in_class = [filename for filename, cid in labels if cid == class_id]


for img_name in images_in_class:
    image_path = os.path.join(folder, img_name)


    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)


    plt.imshow(img, cmap='gray')
    plt.axis('off')
    plt.show()
```

```python
folder = "/content/dataset/train"
output_root = "/content/drive/MyDrive/dataset_by_class"
os.makedirs(output_root, exist_ok=True)
```

```python

all_files = [f for f in os.listdir(folder) if f.endswith(".jpg")]

def get_base(filename):
    return ''.join([c for c in filename if not c.isdigit()]).replace('_','').replace('.jpg','')

unique_bases = list(set(map(get_base, all_files)))

for base in unique_bases:
    class_folder = os.path.join(output_root, base)
    os.makedirs(class_folder, exist_ok=True)

for file in all_files:
    base = get_base(file)
    src = os.path.join(folder, file)
    dst = os.path.join(output_root, base, file)
    shutil.copy(src, dst)

print("✔ Done! All class folders created inside Google Drive.")
```
